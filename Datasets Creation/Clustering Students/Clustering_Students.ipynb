````python
# =============================================================================
# ACADEMIC STRUCTURE CONFIGURATION
# =============================================================================
SPECIALTIES_BY_YEAR = {
    '1': ['A'],  # First year - General studies
    '2': ['A', 'P'],  # Second year - General and Preparatory
    '3': ['A', 'B'],  # Third year - Advanced tracks
    '4': ['DS', 'ARCTIC', 'SIM', 'TWIN', 'NIDS', 'INFINI', 'BI', 'SAE', 'GAMIX', 'IOSYS', 'SE', 'SLEAM'],
    '5': ['DS', 'ARCTIC', 'SIM', 'TWIN', 'NIDS', 'INFINI', 'BI', 'SAE', 'GAMIX', 'IOSYS', 'SE', 'SLEAM']
}

def generate_class_name(year):
    """Generate realistic class names with proper academic formatting"""
    specialty = random.choice(SPECIALTIES_BY_YEAR[year])
    class_number = random.randint(1, 99)
    return f'{year}{specialty}{class_number:02d}'

# Target class distribution
TARGET_CLASSES = 85
CLASSES_PER_YEAR = {'1': 15, '2': 15, '3': 15, '4': 20, '5': 20}

# Generate unique class names
all_generated_classes = []
for year, num_classes in CLASSES_PER_YEAR.items():
    year_classes = set()
    while len(year_classes) < num_classes:
        year_classes.add(generate_class_name(year))
    all_generated_classes.extend(list(year_classes))

# Ensure exact target number
while len(all_generated_classes) < TARGET_CLASSES:
    year = random.choice(list(SPECIALTIES_BY_YEAR.keys()))
    all_generated_classes.append(generate_class_name(year))

all_generated_classes = all_generated_classes[:TARGET_CLASSES]

print(f"🏫 Generated {len(all_generated_classes)} unique classes")
print(f"📝 Sample classes: {random.sample(all_generated_classes, 5)}")
````

````python
# =============================================================================
# ENHANCED STUDENT GENERATION WITH SKILL CORRELATIONS
# =============================================================================
print("👥 Generating students with realistic demographics and correlated skills...")

student_data = []
used_names = set()

for student_id in range(NUM_STUDENTS):
    # Generate demographic information
    nationality = np.random.choice(NATIONALITIES, p=NATIONALITY_WEIGHTS)
    name_data = NAMES_BY_NATIONALITY[nationality]
    
    # Ensure unique names (with fallback to prevent infinite loops)
    attempts = 0
    while attempts < 50:
        gender = random.choice(['Male', 'Female'])
        first_name = random.choice(name_data['male_first'] if gender == 'Male' else name_data['female_first'])
        last_name = random.choice(name_data['last'])
        full_name = f"{first_name} {last_name}"
        if full_name not in used_names:
            used_names.add(full_name)
            break
        attempts += 1
    
    if attempts >= 50:  # Fallback to ensure uniqueness
        first_name = f"{first_name}_{student_id}"
    
    # Enhanced realistic age distribution
    age_prob = random.random()
    if age_prob < 0.02:  # 2% mature students
        age = random.randint(26, 35)
    elif age_prob < 0.05:  # 3% very young students
        age = 17
    else:  # 95% normal distribution
        age = int(np.random.normal(20.5, 1.5))
        age = max(18, min(25, age))  # Reasonable bounds
    
    # Generate correlated skills with realistic patterns
    base_skill = np.random.normal(2.5, 0.8)  # Base competency level
    base_skill = max(0.5, min(4.5, base_skill))
    
    # Create meaningful skill correlations
    hard_skills = base_skill + np.random.normal(0, 0.5)
    soft_skills = base_skill + np.random.normal(0, 0.4) + (0.2 if gender == 'Female' else -0.1)
    teamwork = (hard_skills + soft_skills) / 2 + np.random.normal(0, 0.3)
    creativity = base_skill + np.random.normal(0, 0.6)
    
    # Apply realistic bounds and add individual variation
    skills = {
        'hard_skills': max(0, min(5, hard_skills + np.random.normal(0, 0.2))),
        'soft_skills': max(0, min(5, soft_skills + np.random.normal(0, 0.2))),
        'teamwork': max(0, min(5, teamwork + np.random.normal(0, 0.2))),
        'creativity': max(0, min(5, creativity + np.random.normal(0, 0.2)))
    }
    
    student_data.append({
        'student_id': f'STU_{student_id:04d}',
        'first_name': first_name,
        'last_name': last_name,
        'gender': gender,
        'age': age,
        'nationality': nationality,
        **{k: round(v, 2) for k, v in skills.items()}
    })

# Create initial DataFrame
full_df = pd.DataFrame(student_data)
print(f"✅ Generated {len(full_df)} students with enhanced demographics and correlated skills")
````

````python
# =============================================================================
# OPTIMIZED CLASS ASSIGNMENT (25-35 STUDENTS PER CLASS)
# =============================================================================
print("🏫 Assigning students to classes with optimal team formation sizing...")

class_assignments = []
remaining_students = list(range(NUM_STUDENTS))
random.shuffle(remaining_students)

for class_name in all_generated_classes:
    if not remaining_students:
        break
    
    # Determine optimal class size for team formation
    # Prefer sizes that divide well into 5-student teams
    preferred_sizes = [25, 30, 35]  # All divisible by 5
    weights = [0.3, 0.5, 0.2]  # Prefer 30 students (6 teams of 5)
    class_size = np.random.choice(preferred_sizes, p=weights)
    
    # Adjust if insufficient students remaining
    class_size = min(class_size, len(remaining_students))
    
    # Assign students to this class
    class_students = remaining_students[:class_size]
    remaining_students = remaining_students[class_size:]
    
    for student_idx in class_students:
        class_assignments.append({
            'student_id': f'STU_{student_idx:04d}',
            'class_name': class_name
        })

# Handle any remaining students (assign to existing classes)
if remaining_students:
    print(f"📝 Distributing {len(remaining_students)} remaining students...")
    for i, student_idx in enumerate(remaining_students):
        target_class = all_generated_classes[i % len(all_generated_classes)]
        class_assignments.append({
            'student_id': f'STU_{student_idx:04d}',
            'class_name': target_class
        })

# Create assignments DataFrame and merge
class_assignments_df = pd.DataFrame(class_assignments)
final_df = full_df.merge(class_assignments_df, on='student_id', how='left')

# Save the dataset
output_path = os.path.join(BASE_PATH, OUTPUT_FILE)
final_df.to_csv(output_path, index=False)

print(f"💾 Dataset saved to: {output_path}")
print(f"📊 Total students: {len(final_df)}")
print(f"🏫 Total classes: {final_df['class_name'].nunique()}")
print(f"👥 Average class size: {final_df['class_name'].value_counts().mean():.1f}")
print(f"📈 Class size range: {final_df['class_name'].value_counts().min()}-{final_df['class_name'].value_counts().max()}")
````

````python
# =============================================================================
# SMART CLASS SELECTION FOR ANALYSIS
# =============================================================================
print(f"\n🎯 SELECTING OPTIMAL CLASS FOR DETAILED ANALYSIS")

# Filter classes with optimal size for team formation (25-35 students)
optimal_classes = class_summary[
    (class_summary['num_students'] >= 25) & 
    (class_summary['num_students'] <= 35)
].copy()

if len(optimal_classes) > 0:
    # Select class with size closest to 30 (6 teams of 5)
    optimal_classes['distance_from_30'] = abs(optimal_classes['num_students'] - 30)
    selected_class_name = optimal_classes.sort_values('distance_from_30').index[0]
    
    # Extract class data
    class_df = df[df['class_name'] == selected_class_name].copy().reset_index(drop=True)
    
    print(f"✅ Selected class: {selected_class_name}")
    print(f"👥 Students in class: {len(class_df)}")
    print(f"🎯 Optimal for: {len(class_df)//5} teams of 5 students each")
    
    # Display class characteristics
    print(f"\n📊 CLASS CHARACTERISTICS:")
    print(f"   Age range: {class_df['age'].min()}-{class_df['age'].max()} years (avg: {class_df['age'].mean():.1f})")
    print(f"   Gender distribution: {class_df['gender'].value_counts().to_dict()}")
    print(f"   Nationalities: {class_df['nationality'].nunique()} different ({', '.join(class_df['nationality'].unique()[:3])}...)")
    
    skill_stats = class_df[features].describe().round(2)
    print(f"\n🔧 SKILL DISTRIBUTION:")
    display(skill_stats.loc[['mean', 'std', 'min', 'max']])
````

### 🎯 **Clustering Optimization Theory**

#### **📈 K-Means Mathematical Foundation**

The K-Means algorithm optimizes the **Within-Cluster Sum of Squares (WCSS)**:

```
Objective Function: minimize Σᵢ₌₁ᵏ Σₓ∈Cᵢ ||x - μᵢ||²

Where:
- k = number of clusters (teams)
- Cᵢ = cluster i
- μᵢ = centroid of cluster i
- ||x - μᵢ||² = squared Euclidean distance
```

#### **🔬 Optimal K Selection Methodology**

We employ **multiple convergent approaches** for robust K selection:

1. **Elbow Method**: Identifies the "elbow" in WCSS reduction curve
2. **Silhouette Analysis**: Maximizes average silhouette score
3. **Calinski-Harabasz**: Maximizes ratio of between/within cluster variance
4. **Davies-Bouldin**: Minimizes average cluster similarity
5. **Composite Scoring**: Weighted combination of all metrics

#### **🎲 Initialization Strategy**

**K-Means++ Initialization** provides superior convergence:
- **Smart Seeding**: First centroid chosen randomly
- **Distance-Based Selection**: Subsequent centroids chosen with probability proportional to squared distance
- **Convergence Guarantee**: Typically converges faster than random initialization
- **Multiple Runs**: 10 initializations with best result selection

#### **📊 Convergence Criteria**

```python

tolerance = 1e-4  # Minimum centroid movement
max_iterations = 300  # Maximum algorithm iterations
```

````python
# 📊 COMPREHENSIVE SILHOUETTE ANALYSIS & VISUALIZATION
print("="*80)
print("📈 DETAILED SILHOUETTE ANALYSIS & CLUSTER QUALITY ASSESSMENT")
print("="*80)

def detailed_silhouette_analysis(X_scaled, training_results, optimal_k, class_labels, verbose=True):
    """
    Comprehensive silhouette analysis with detailed visualizations and insights
    """
    
    if verbose:
        print(f"\n🔍 SILHOUETTE ANALYSIS OVERVIEW:")
        print(f"   • Optimal K: {optimal_k}")
        print(f"   • Data Points: {len(X_scaled)}")
        print(f"   • Features: {X_scaled.shape[1]}")
    
    # Calculate detailed silhouette metrics for optimal k
    silhouette_avg = silhouette_score(X_scaled, class_labels)
    sample_silhouette_values = silhouette_samples(X_scaled, class_labels)
    
    if verbose:
        print(f"\n📊 SILHOUETTE SCORE BREAKDOWN:")
        print(f"   • Average Silhouette Score: {silhouette_avg:.4f}")
        print(f"   • Score Interpretation:")
        if silhouette_avg >= 0.7:
            print(f"     🟢 EXCELLENT clustering (≥0.7)")
        elif silhouette_avg >= 0.5:
            print(f"     🟡 GOOD clustering (0.5-0.7)")
        elif silhouette_avg >= 0.25:
            print(f"     🟠 FAIR clustering (0.25-0.5)")
        else:
            print(f"     🔴 POOR clustering (<0.25)")
    
    # Analyze silhouette scores by cluster
    cluster_silhouette_stats = {}
    unique_labels = np.unique(class_labels)
    
    print(f"\n📋 PER-CLUSTER SILHOUETTE ANALYSIS:")
    for cluster_id in unique_labels:
        cluster_mask = (class_labels == cluster_id)
        cluster_silhouette_scores = sample_silhouette_values[cluster_mask]
        
        cluster_stats = {
            'count': len(cluster_silhouette_scores),
            'mean': np.mean(cluster_silhouette_scores),
            'std': np.std(cluster_silhouette_scores),
            'min': np.min(cluster_silhouette_scores),
            'max': np.max(cluster_silhouette_scores),
            'median': np.median(cluster_silhouette_scores),
            'negative_count': np.sum(cluster_silhouette_scores < 0),
            'excellent_count': np.sum(cluster_silhouette_scores >= 0.7),
            'good_count': np.sum((cluster_silhouette_scores >= 0.5) & (cluster_silhouette_scores < 0.7)),
            'scores': cluster_silhouette_scores
        }
        
        cluster_silhouette_stats[cluster_id] = cluster_stats
        
        if verbose:
            print(f"\n   🎯 Cluster {cluster_id} ({cluster_stats['count']} students):")
            print(f"     • Mean Silhouette: {cluster_stats['mean']:.4f}")
            print(f"     • Std Deviation: {cluster_stats['std']:.4f}")
            print(f"     • Range: [{cluster_stats['min']:.4f}, {cluster_stats['max']:.4f}]")
            print(f"     • Median: {cluster_stats['median']:.4f}")
            print(f"     • Negative Scores: {cluster_stats['negative_count']} ({cluster_stats['negative_count']/cluster_stats['count']*100:.1f}%)")
            print(f"     • Excellent (≥0.7): {cluster_stats['excellent_count']} ({cluster_stats['excellent_count']/cluster_stats['count']*100:.1f}%)")
            print(f"     • Good (0.5-0.7): {cluster_stats['good_count']} ({cluster_stats['good_count']/cluster_stats['count']*100:.1f}%)")
    
    # Overall quality metrics
    total_negative = np.sum(sample_silhouette_values < 0)
    total_excellent = np.sum(sample_silhouette_values >= 0.7)
    total_good = np.sum((sample_silhouette_values >= 0.5) & (sample_silhouette_values < 0.7))
    total_fair = np.sum((sample_silhouette_values >= 0.25) & (sample_silhouette_values < 0.5))
    total_poor = np.sum(sample_silhouette_values < 0.25)
    
    if verbose:
        print(f"\n🏆 OVERALL QUALITY DISTRIBUTION:")
        print(f"   • Excellent (≥0.7): {total_excellent} students ({total_excellent/len(sample_silhouette_values)*100:.1f}%)")
        print(f"   • Good (0.5-0.7): {total_good} students ({total_good/len(sample_silhouette_values)*100:.1f}%)")
        print(f"   • Fair (0.25-0.5): {total_fair} students ({total_fair/len(sample_silhouette_values)*100:.1f}%)")
        print(f"   • Poor (<0.25): {total_poor} students ({total_poor/len(sample_silhouette_values)*100:.1f}%)")
        print(f"   • Misclassified (<0): {total_negative} students ({total_negative/len(sample_silhouette_values)*100:.1f}%)")
    
    # Create comprehensive silhouette visualization
    from plotly.subplots import make_subplots
    
    fig = make_subplots(
        rows=2, cols=3,
        subplot_titles=[
            'Silhouette Scores by K',
            'Calinski-Harabasz Scores',
            'Davies-Bouldin Scores',
            'Silhouette Distribution (Optimal K)',
            'Cluster Quality Comparison',
            'Individual Silhouette Scores'
        ],
        specs=[[{"secondary_y": False}, {"secondary_y": False}, {"secondary_y": False}],
               [{"secondary_y": False}, {"secondary_y": False}, {"secondary_y": False}]]
    )
    
    # Plot 1: Silhouette scores by K
    fig.add_trace(go.Scatter(
        x=training_results['k_values'],
        y=training_results['silhouette_avg'],
        mode='lines+markers',
        name='Silhouette Score',
        line=dict(color='blue', width=3),
        marker=dict(size=8)
    ), row=1, col=1)
    
    # Highlight optimal k
    optimal_k_idx = training_results['k_values'].index(optimal_k)
    fig.add_trace(go.Scatter(
        x=[optimal_k],
        y=[training_results['silhouette_avg'][optimal_k_idx]],
        mode='markers',
        name=f'Optimal K={optimal_k}',
        marker=dict(color='red', size=12, symbol='star')
    ), row=1, col=1)
    
    # Plot 2: Calinski-Harabasz scores
    fig.add_trace(go.Scatter(
        x=training_results['k_values'],
        y=training_results['calinski_harabasz'],
        mode='lines+markers',
        name='Calinski-Harabasz',
        line=dict(color='green', width=3),
        marker=dict(size=8)
    ), row=1, col=2)
    
    # Plot 3: Davies-Bouldin scores
    fig.add_trace(go.Scatter(
        x=training_results['k_values'],
        y=training_results['davies_bouldin'],
        mode='lines+markers',
        name='Davies-Bouldin',
        line=dict(color='orange', width=3),
        marker=dict(size=8)
    ), row=1, col=3)
    
    # Plot 4: Silhouette distribution for optimal k
    fig.add_trace(go.Histogram(
        x=sample_silhouette_values,
        nbinsx=20,
        name='Silhouette Distribution',
        marker=dict(color='skyblue', opacity=0.7)
    ), row=2, col=1)
    
    # Plot 5: Cluster quality comparison
    cluster_means = [cluster_silhouette_stats[cid]['mean'] for cid in unique_labels]
    cluster_names = [f'Cluster {cid}' for cid in unique_labels]
    
    fig.add_trace(go.Bar(
        x=cluster_names,
        y=cluster_means,
        name='Mean Silhouette by Cluster',
        marker=dict(color='lightgreen', opacity=0.8)
    ), row=2, col=2)
    
    # Plot 6: Individual silhouette scores
    y_lower = 10
    colors = ['red', 'blue', 'green', 'orange', 'purple', 'brown', 'pink', 'gray']
    
    for i, cluster_id in enumerate(unique_labels):
        cluster_silhouette_scores = cluster_silhouette_stats[cluster_id]['scores']
        cluster_silhouette_scores.sort()
        
        size_cluster_i = len(cluster_silhouette_scores)
        y_upper = y_lower + size_cluster_i
        
        color = colors[i % len(colors)]
        fig.add_trace(go.Scatter(
            x=cluster_silhouette_scores,
            y=list(range(y_lower, y_upper)),
            mode='markers',
            name=f'Cluster {cluster_id}',
            marker=dict(color=color, size=4),
            showlegend=False
        ), row=2, col=3)
        
        y_lower = y_upper + 10
    
    # Add vertical line for average silhouette score
    fig.add_vline(x=silhouette_avg, line_dash="dash", line_color="red", 
                  annotation_text=f"Avg: {silhouette_avg:.3f}", row=2, col=3)
    
    # Update layout
    fig.update_layout(
        title_text="Comprehensive Silhouette Analysis Results",
        height=800,
        showlegend=True
    )
    
    # Update axis labels
    fig.update_xaxes(title_text="Number of Clusters (K)", row=1, col=1)
    fig.update_yaxes(title_text="Silhouette Score", row=1, col=1)
    
    fig.update_xaxes(title_text="Number of Clusters (K)", row=1, col=2)
    fig.update_yaxes(title_text="Calinski-Harabasz Score", row=1, col=2)
    
    fig.update_xaxes(title_text="Number of Clusters (K)", row=1, col=3)
    fig.update_yaxes(title_text="Davies-Bouldin Score", row=1, col=3)
    
    fig.update_xaxes(title_text="Silhouette Score", row=2, col=1)
    fig.update_yaxes(title_text="Frequency", row=2, col=1)
    
    fig.update_xaxes(title_text="Clusters", row=2, col=2)
    fig.update_yaxes(title_text="Mean Silhouette Score", row=2, col=2)
    
    fig.update_xaxes(title_text="Silhouette Score", row=2, col=3)
    fig.update_yaxes(title_text="Student Index", row=2, col=3)
    
    fig.show()
    
    # Statistical significance testing
    if verbose:
        print(f"\n📊 STATISTICAL VALIDATION:")
        
        # Test cluster separation
        from scipy import stats
        cluster_pairs_tests = []
        for i in range(len(unique_labels)):
            for j in range(i+1, len(unique_labels)):
                cluster_i_scores = cluster_silhouette_stats[unique_labels[i]]['scores']
                cluster_j_scores = cluster_silhouette_stats[unique_labels[j]]['scores']
                
                # Mann-Whitney U test for cluster separation
                statistic, p_value = stats.mannwhitneyu(cluster_i_scores, cluster_j_scores, 
                                                       alternative='two-sided')
                cluster_pairs_tests.append((i, j, p_value))
                
                if p_value < 0.05:
                    significance = "Significantly different"
                else:
                    significance = "Not significantly different"
                
                print(f"   • Cluster {unique_labels[i]} vs {unique_labels[j]}: p={p_value:.4f} ({significance})")
        
        # Overall clustering quality assessment
        quality_score = (total_excellent * 1.0 + total_good * 0.8 + total_fair * 0.5 + 
                        total_poor * 0.2 - total_negative * 0.5) / len(sample_silhouette_values)
        
        print(f"\n🏆 FINAL CLUSTERING QUALITY ASSESSMENT:")
        print(f"   • Overall Quality Score: {quality_score:.4f}")
        print(f"   • Clustering Effectiveness: {silhouette_avg:.4f}")
        print(f"   • Cluster Separation Quality: {'Strong' if np.mean([p < 0.05 for _, _, p in cluster_pairs_tests]) > 0.7 else 'Moderate'}")
        
        if quality_score >= 0.8:
            print(f"   • 🟢 RECOMMENDATION: Excellent clustering - Proceed with confidence")
        elif quality_score >= 0.6:
            print(f"   • 🟡 RECOMMENDATION: Good clustering - Suitable for most applications")
        elif quality_score >= 0.4:
            print(f"   • 🟠 RECOMMENDATION: Fair clustering - Consider parameter tuning")
        else:
            print(f"   • 🔴 RECOMMENDATION: Poor clustering - Requires algorithm revision")
    
    return cluster_silhouette_stats, quality_score

# Execute detailed silhouette analysis
print(f"🚀 Starting comprehensive silhouette analysis...")
silhouette_stats, quality_score = detailed_silhouette_analysis(
    X_scaled, training_results, optimal_k, class_labels, verbose=True
)

print(f"\n🎉 SILHOUETTE ANALYSIS COMPLETED!")
print(f"="*80)
````

````python
# 🔬 ADVANCED ALGORITHM COMPARISON & CROSS-VALIDATION
print("="*80)
print("🧪 COMPREHENSIVE ALGORITHM TRAINING & PERFORMANCE COMPARISON")
print("="*80)

# Ensure all necessary imports
from sklearn.cluster import AgglomerativeClustering
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score, silhouette_samples
import time
import numpy as np

def comprehensive_algorithm_comparison(X_scaled, optimal_k, verbose=True):
    """
    Train and compare multiple clustering algorithms with comprehensive metrics
    """
    
    algorithms = {
        'K-Means': KMeans(n_clusters=optimal_k, init='k-means++', n_init=10, random_state=42),
        'K-Means (Random Init)': KMeans(n_clusters=optimal_k, init='random', n_init=10, random_state=42),
        'Agglomerative (Ward)': AgglomerativeClustering(n_clusters=optimal_k, linkage='ward'),
        'Agglomerative (Complete)': AgglomerativeClustering(n_clusters=optimal_k, linkage='complete'),
        'Agglomerative (Average)': AgglomerativeClustering(n_clusters=optimal_k, linkage='average'),
        'Gaussian Mixture': GaussianMixture(n_components=optimal_k, random_state=42)
    }
    
    results = {}
    
    if verbose:
        print(f"\n🎯 TRAINING {len(algorithms)} DIFFERENT ALGORITHMS:")
        print(f"   • Target Clusters: {optimal_k}")
        print(f"   • Dataset Size: {X_scaled.shape}")
        print(f"   • Evaluation Metrics: Silhouette, Calinski-Harabasz, Davies-Bouldin")
    
    for alg_name, algorithm in algorithms.items():
        if verbose:
            print(f"\n🔄 Training {alg_name}...")
        
        start_time = time.time()
        
        try:
            # Fit the algorithm
            if hasattr(algorithm, 'fit_predict'):
                labels = algorithm.fit_predict(X_scaled)
            else:
                algorithm.fit(X_scaled)
                if hasattr(algorithm, 'labels_'):
                    labels = algorithm.labels_
                else:
                    labels = algorithm.predict(X_scaled)
            
            training_time = time.time() - start_time
            
            # Calculate metrics
            silhouette_avg = silhouette_score(X_scaled, labels)
            calinski_harabasz = calinski_harabasz_score(X_scaled, labels)
            davies_bouldin = davies_bouldin_score(X_scaled, labels)
            
            # Additional metrics
            unique_labels, counts = np.unique(labels, return_counts=True)
            cluster_balance = np.std(counts) / np.mean(counts) if np.mean(counts) > 0 else 0  # Coefficient of variation
            cluster_sizes = list(counts)
            
            # Silhouette per cluster
            silhouette_samples_scores = silhouette_samples(X_scaled, labels)
            cluster_silhouette_means = []
            for cluster_id in unique_labels:
                mask = (labels == cluster_id)
                cluster_sil_mean = np.mean(silhouette_samples_scores[mask])
                cluster_silhouette_means.append(cluster_sil_mean)
            
            silhouette_consistency = np.std(cluster_silhouette_means)
            
            # Store results
            results[alg_name] = {
                'algorithm': algorithm,
                'labels': labels,
                'training_time': training_time,
                'silhouette_score': silhouette_avg,
                'calinski_harabasz': calinski_harabasz,
                'davies_bouldin': davies_bouldin,
                'cluster_balance': cluster_balance,
                'cluster_sizes': cluster_sizes,
                'cluster_silhouette_means': cluster_silhouette_means,
                'silhouette_consistency': silhouette_consistency,
                'n_clusters': len(unique_labels)
            }
            
            if verbose:
                print(f"   ✅ {alg_name} Training Complete:")
                print(f"     ⏱️  Time: {training_time:.4f}s")
                print(f"     📊 Silhouette: {silhouette_avg:.4f}")
                print(f"     📈 Calinski-Harabasz: {calinski_harabasz:.2f}")
                print(f"     📉 Davies-Bouldin: {davies_bouldin:.4f}")
                print(f"     ⚖️  Cluster Balance (CV): {cluster_balance:.4f}")
                print(f"     📋 Cluster Sizes: {cluster_sizes}")
                print(f"     🎯 Silhouette Consistency: {silhouette_consistency:.4f}")
                
        except Exception as e:
            if verbose:
                print(f"   ❌ {alg_name} Failed: {str(e)}")
            results[alg_name] = None
    
    # Filter successful results
    successful_results = {k: v for k, v in results.items() if v is not None}
    
    if len(successful_results) == 0:
        print("⚠️ No algorithms completed successfully!")
        return {}, {}, ('None', 0)
    
    if verbose:
        print(f"\n📊 ALGORITHM PERFORMANCE RANKING:")
        
        # Ranking by different metrics
        metrics_rankings = {}
        
        # Silhouette ranking (higher is better)
        sil_sorted = sorted(successful_results.items(), 
                           key=lambda x: x[1]['silhouette_score'], reverse=True)
        metrics_rankings['silhouette'] = sil_sorted
        
        # Calinski-Harabasz ranking (higher is better)
        cal_sorted = sorted(successful_results.items(), 
                           key=lambda x: x[1]['calinski_harabasz'], reverse=True)
        metrics_rankings['calinski_harabasz'] = cal_sorted
        
        # Davies-Bouldin ranking (lower is better)
        dav_sorted = sorted(successful_results.items(), 
                           key=lambda x: x[1]['davies_bouldin'])
        metrics_rankings['davies_bouldin'] = dav_sorted
        
        # Balance ranking (lower is better)
        bal_sorted = sorted(successful_results.items(), 
                           key=lambda x: x[1]['cluster_balance'])
        metrics_rankings['balance'] = bal_sorted
        
        # Speed ranking (lower is better)
        speed_sorted = sorted(successful_results.items(), 
                             key=lambda x: x[1]['training_time'])
        metrics_rankings['speed'] = speed_sorted
        
        # Display rankings
        for metric, ranking in metrics_rankings.items():
            print(f"\n   🏆 {metric.replace('_', ' ').title()} Ranking:")
            for i, (alg_name, result) in enumerate(ranking[:3]):  # Top 3
                if metric == 'silhouette':
                    value = result['silhouette_score']
                elif metric == 'calinski_harabasz':
                    value = result['calinski_harabasz']
                elif metric == 'davies_bouldin':
                    value = result['davies_bouldin']
                elif metric == 'balance':
                    value = result['cluster_balance']
                else:  # speed
                    value = result['training_time']
                print(f"     {i+1}. {alg_name}: {value:.4f}")
        
        # Composite scoring
        print(f"\n🎯 COMPOSITE PERFORMANCE SCORING:")
        composite_scores = {}
        
        for alg_name, result in successful_results.items():
            # Normalize metrics (0-1 scale)
            sil_norm = result['silhouette_score']  # Already 0-1
            
            # Normalize Calinski-Harabasz (min-max scaling)
            cal_values = [r['calinski_harabasz'] for r in successful_results.values()]
            if max(cal_values) > min(cal_values):
                cal_norm = (result['calinski_harabasz'] - min(cal_values)) / (max(cal_values) - min(cal_values))
            else:
                cal_norm = 1.0
            
            # Normalize Davies-Bouldin (inverse and min-max)
            dav_values = [r['davies_bouldin'] for r in successful_results.values()]
            if max(dav_values) > min(dav_values):
                dav_norm = 1 - (result['davies_bouldin'] - min(dav_values)) / (max(dav_values) - min(dav_values))
            else:
                dav_norm = 1.0
            
            # Normalize balance (inverse and min-max)
            bal_values = [r['cluster_balance'] for r in successful_results.values()]
            if max(bal_values) > min(bal_values):
                bal_norm = 1 - (result['cluster_balance'] - min(bal_values)) / (max(bal_values) - min(bal_values))
            else:
                bal_norm = 1.0
            
            # Composite score (weighted average)
            composite = (sil_norm * 0.4 + cal_norm * 0.3 + dav_norm * 0.2 + bal_norm * 0.1)
            composite_scores[alg_name] = composite
            
            print(f"   • {alg_name}: {composite:.4f}")
            print(f"     - Silhouette: {sil_norm:.3f} (weight: 0.4)")
            print(f"     - Calinski-H: {cal_norm:.3f} (weight: 0.3)")
            print(f"     - Davies-B: {dav_norm:.3f} (weight: 0.2)")
            print(f"     - Balance: {bal_norm:.3f} (weight: 0.1)")
        
        # Best algorithm
        if composite_scores:
            best_algorithm = max(composite_scores.items(), key=lambda x: x[1])
            print(f"\n🏆 BEST ALGORITHM: {best_algorithm[0]} (Score: {best_algorithm[1]:.4f})")
        else:
            best_algorithm = ('None', 0)
    
    return successful_results, composite_scores, best_algorithm

# Execute algorithm comparison
print(f"🚀 Starting comprehensive algorithm comparison...")
algorithm_results, composite_scores, best_algorithm = comprehensive_algorithm_comparison(
    X_scaled, optimal_k, verbose=True
)

# Create comparison visualization only if we have results
if algorithm_results:
    # Create comparison visualization
    fig = make_subplots(
        rows=2, cols=2,
        subplot_titles=[
            'Silhouette Score Comparison',
            'Training Time Comparison', 
            'Cluster Balance Comparison',
            'Composite Score Ranking'
        ]
    )

    # Extract data for plotting
    alg_names = list(algorithm_results.keys())
    silhouette_scores = [algorithm_results[name]['silhouette_score'] for name in alg_names]
    training_times = [algorithm_results[name]['training_time'] for name in alg_names]
    balance_scores = [algorithm_results[name]['cluster_balance'] for name in alg_names]
    composite_values = [composite_scores[name] for name in alg_names]

    # Plot 1: Silhouette scores
    fig.add_trace(go.Bar(
        x=alg_names,
        y=silhouette_scores,
        name='Silhouette Score',
        marker_color='lightblue'
    ), row=1, col=1)

    # Plot 2: Training times
    fig.add_trace(go.Bar(
        x=alg_names,
        y=training_times,
        name='Training Time (s)',
        marker_color='lightgreen'
    ), row=1, col=2)

    # Plot 3: Balance scores
    fig.add_trace(go.Bar(
        x=alg_names,
        y=balance_scores,
        name='Cluster Balance',
        marker_color='lightcoral'
    ), row=2, col=1)

    # Plot 4: Composite scores
    fig.add_trace(go.Bar(
        x=alg_names,
        y=composite_values,
        name='Composite Score',
        marker_color='gold'
    ), row=2, col=2)

    fig.update_layout(
        title_text="Comprehensive Algorithm Performance Comparison",
        height=800,
        showlegend=False
    )

    # Rotate x-axis labels for better readability
    fig.update_xaxes(tickangle=45)

    fig.show()

print(f"\n🎉 ALGORITHM COMPARISON COMPLETED!")
if best_algorithm[0] != 'None':
    print(f"📊 Best performing algorithm: {best_algorithm[0]}")
else:
    print(f"⚠️ No algorithms completed successfully")
print(f"="*80)
````

````python
# 🔬 STATISTICAL VALIDATION & CROSS-VALIDATION ANALYSIS
print("="*80)
print("📊 STATISTICAL VALIDATION & ROBUSTNESS TESTING")
print("="*80)

def statistical_validation_analysis(X_scaled, optimal_k, best_algorithm_name, algorithm_results, verbose=True):
    """
    Perform comprehensive statistical validation and robustness testing
    """
    
    if verbose:
        print(f"\n🔍 STATISTICAL VALIDATION SETUP:")
        print(f"   • Best Algorithm: {best_algorithm_name}")
        print(f"   • Optimal K: {optimal_k}")
        print(f"   • Validation Methods: Bootstrap, Stability, Consistency")
    
    # Get the best algorithm result
    best_result = algorithm_results[best_algorithm_name]
    best_labels = best_result['labels']
    
    # 1. Bootstrap Validation
    if verbose:
        print(f"\n🔄 BOOTSTRAP VALIDATION (100 iterations):")
    
    bootstrap_results = {
        'silhouette_scores': [],
        'calinski_harabasz_scores': [],
        'davies_bouldin_scores': [],
        'cluster_counts': []
    }
    
    n_bootstrap = 100
    n_samples = len(X_scaled)
    
    for i in range(n_bootstrap):
        # Bootstrap sampling
        bootstrap_indices = np.random.choice(n_samples, size=n_samples, replace=True)
        X_bootstrap = X_scaled[bootstrap_indices]
        
        try:
            # Fit the best algorithm on bootstrap sample
            if best_algorithm_name.startswith('K-Means'):
                model = KMeans(n_clusters=optimal_k, init='k-means++', n_init=5, random_state=i)
            elif best_algorithm_name.startswith('Agglomerative'):
                if 'Ward' in best_algorithm_name:
                    model = AgglomerativeClustering(n_clusters=optimal_k, linkage='ward')
                elif 'Complete' in best_algorithm_name:
                    model = AgglomerativeClustering(n_clusters=optimal_k, linkage='complete')
                else:
                    model = AgglomerativeClustering(n_clusters=optimal_k, linkage='average')
            else:  # Gaussian Mixture
                model = GaussianMixture(n_components=optimal_k, random_state=i)
            
            bootstrap_labels = model.fit_predict(X_bootstrap)
            
            # Calculate metrics
            sil_score = silhouette_score(X_bootstrap, bootstrap_labels)
            cal_score = calinski_harabasz_score(X_bootstrap, bootstrap_labels)
            dav_score = davies_bouldin_score(X_bootstrap, bootstrap_labels)
            n_clusters = len(np.unique(bootstrap_labels))
            
            bootstrap_results['silhouette_scores'].append(sil_score)
            bootstrap_results['calinski_harabasz_scores'].append(cal_score)
            bootstrap_results['davies_bouldin_scores'].append(dav_score)
            bootstrap_results['cluster_counts'].append(n_clusters)
            
        except Exception as e:
            continue  # Skip failed iterations
        
        if verbose and (i + 1) % 20 == 0:
            print(f"     Completed {i + 1}/{n_bootstrap} bootstrap iterations...")
    
    # Bootstrap statistics
    if bootstrap_results['silhouette_scores']:
        sil_mean = np.mean(bootstrap_results['silhouette_scores'])
        sil_std = np.std(bootstrap_results['silhouette_scores'])
        sil_ci_lower = np.percentile(bootstrap_results['silhouette_scores'], 2.5)
        sil_ci_upper = np.percentile(bootstrap_results['silhouette_scores'], 97.5)
        
        cal_mean = np.mean(bootstrap_results['calinski_harabasz_scores'])
        cal_std = np.std(bootstrap_results['calinski_harabasz_scores'])
        
        dav_mean = np.mean(bootstrap_results['davies_bouldin_scores'])
        dav_std = np.std(bootstrap_results['davies_bouldin_scores'])
        
        cluster_stability = np.mean([c == optimal_k for c in bootstrap_results['cluster_counts']])
        
        if verbose:
            print(f"\n📊 BOOTSTRAP RESULTS ({len(bootstrap_results['silhouette_scores'])} successful iterations):")
            print(f"   • Silhouette Score:")
            print(f"     - Mean: {sil_mean:.4f} ± {sil_std:.4f}")
            print(f"     - 95% CI: [{sil_ci_lower:.4f}, {sil_ci_upper:.4f}]")
            print(f"     - Original: {best_result['silhouette_score']:.4f}")
            print(f"   • Calinski-Harabasz: {cal_mean:.2f} ± {cal_std:.2f}")
            print(f"   • Davies-Bouldin: {dav_mean:.4f} ± {dav_std:.4f}")
            print(f"   • Cluster Stability: {cluster_stability:.1%}")
    
    # 2. Stability Analysis with Different Random Seeds
    if verbose:
        print(f"\n🎯 STABILITY ANALYSIS (50 random seeds):")
    
    stability_results = {
        'silhouette_scores': [],
        'adjusted_rand_scores': [],
        'normalized_mutual_info_scores': []
    }
    
    reference_labels = best_labels.copy()
    
    for seed in range(50):
        try:
            if best_algorithm_name.startswith('K-Means'):
                model = KMeans(n_clusters=optimal_k, init='k-means++', n_init=10, random_state=seed)
            elif best_algorithm_name.startswith('Agglomerative'):
                # Agglomerative is deterministic, so we'll use different linkage criteria
                continue
            else:  # Gaussian Mixture
                model = GaussianMixture(n_components=optimal_k, random_state=seed)
            
            test_labels = model.fit_predict(X_scaled)
            
            # Calculate stability metrics
            sil_score = silhouette_score(X_scaled, test_labels)
            ari_score = adjusted_rand_score(reference_labels, test_labels)
            nmi_score = normalized_mutual_info_score(reference_labels, test_labels)
            
            stability_results['silhouette_scores'].append(sil_score)
            stability_results['adjusted_rand_scores'].append(ari_score)
            stability_results['normalized_mutual_info_scores'].append(nmi_score)
            
        except Exception as e:
            continue
    
    if stability_results['silhouette_scores']:
        sil_stability = np.std(stability_results['silhouette_scores'])
        ari_mean = np.mean(stability_results['adjusted_rand_scores'])
        nmi_mean = np.mean(stability_results['normalized_mutual_info_scores'])
        
        if verbose:
            print(f"   • Silhouette Stability (std): {sil_stability:.4f}")
            print(f"   • Mean Adjusted Rand Index: {ari_mean:.4f}")
            print(f"   • Mean Normalized Mutual Info: {nmi_mean:.4f}")
            print(f"   • Algorithm Consistency: {'High' if sil_stability < 0.05 else 'Moderate' if sil_stability < 0.1 else 'Low'}")
    
    # 3. Cluster Validation Metrics
    if verbose:
        print(f"\n🔍 ADVANCED CLUSTER VALIDATION:")
    
    # Inertia analysis for K-Means
    if best_algorithm_name.startswith('K-Means'):
        kmeans_model = algorithm_results[best_algorithm_name]['algorithm']
        total_inertia = kmeans_model.inertia_
        
        # Calculate within-cluster sum of squares for each cluster
        cluster_inertias = []
        for cluster_id in range(optimal_k):
            cluster_mask = (best_labels == cluster_id)
            if np.sum(cluster_mask) > 0:
                cluster_data = X_scaled[cluster_mask]
                cluster_center = kmeans_model.cluster_centers_[cluster_id]
                cluster_inertia = np.sum((cluster_data - cluster_center) ** 2)
                cluster_inertias.append(cluster_inertia)
        
        inertia_balance = np.std(cluster_inertias) / np.mean(cluster_inertias) if cluster_inertias else 0
        
        if verbose:
            print(f"   • Total Inertia: {total_inertia:.2f}")
            print(f"   • Cluster Inertias: {[f'{x:.2f}' for x in cluster_inertias]}")
            print(f"   • Inertia Balance (CV): {inertia_balance:.4f}")
    
    # Gap statistic approximation
    if verbose:
        print(f"\n📈 CLUSTERING QUALITY ASSESSMENT:")
    
    # Calculate inter-cluster distances
    cluster_centers = []
    if best_algorithm_name.startswith('K-Means'):
        cluster_centers = algorithm_results[best_algorithm_name]['algorithm'].cluster_centers_
    else:
        # Calculate centroids manually
        for cluster_id in range(optimal_k):
            cluster_mask = (best_labels == cluster_id)
            if np.sum(cluster_mask) > 0:
                cluster_center = np.mean(X_scaled[cluster_mask], axis=0)
                cluster_centers.append(cluster_center)
    
    if len(cluster_centers) > 1:
        # Calculate pairwise distances between cluster centers
        from scipy.spatial.distance import pdist
        center_distances = pdist(cluster_centers)
        min_center_distance = np.min(center_distances)
        max_center_distance = np.max(center_distances)
        avg_center_distance = np.mean(center_distances)
        
        # Calculate average within-cluster distance
        within_cluster_distances = []
        for cluster_id in range(optimal_k):
            cluster_mask = (best_labels == cluster_id)
            cluster_data = X_scaled[cluster_mask]
            if len(cluster_data) > 1:
                cluster_distances = pdist(cluster_data)
                within_cluster_distances.extend(cluster_distances)
        
        avg_within_distance = np.mean(within_cluster_distances) if within_cluster_distances else 0
        
        # Separation ratio
        separation_ratio = avg_center_distance / (avg_within_distance + 1e-8)
        
        if verbose:
            print(f"   • Inter-cluster Separation:")
            print(f"     - Min distance: {min_center_distance:.4f}")
            print(f"     - Max distance: {max_center_distance:.4f}")
            print(f"     - Avg distance: {avg_center_distance:.4f}")
            print(f"   • Avg Within-cluster Distance: {avg_within_distance:.4f}")
            print(f"   • Separation Ratio: {separation_ratio:.4f}")
            print(f"   • Cluster Quality: {'Excellent' if separation_ratio > 2 else 'Good' if separation_ratio > 1.5 else 'Fair'}")
    
    # Final validation summary
    validation_summary = {
        'bootstrap_silhouette_mean': sil_mean if 'sil_mean' in locals() else None,
        'bootstrap_silhouette_std': sil_std if 'sil_std' in locals() else None,
        'stability_score': sil_stability if 'sil_stability' in locals() else None,
        'cluster_stability': cluster_stability if 'cluster_stability' in locals() else None,
        'separation_ratio': separation_ratio if 'separation_ratio' in locals() else None
    }
    
    if verbose:
        print(f"\n🏆 VALIDATION SUMMARY:")
        overall_quality = "Excellent"
        if validation_summary['bootstrap_silhouette_mean'] and validation_summary['bootstrap_silhouette_mean'] < 0.5:
            overall_quality = "Fair"
        elif validation_summary['stability_score'] and validation_summary['stability_score'] > 0.1:
            overall_quality = "Moderate"
        
        print(f"   • Overall Quality: {overall_quality}")
        print(f"   • Recommended for Production: {'Yes' if overall_quality in ['Excellent', 'Good'] else 'With Caution'}")
        print(f"   • Confidence Level: {'High' if overall_quality == 'Excellent' else 'Medium' if overall_quality == 'Good' else 'Low'}")
    
    return validation_summary, bootstrap_results, stability_results

# Execute statistical validation
print(f"🚀 Starting statistical validation analysis...")
validation_summary, bootstrap_results, stability_results = statistical_validation_analysis(
    X_scaled, optimal_k, best_algorithm[0], algorithm_results, verbose=True
)

print(f"\n🎉 STATISTICAL VALIDATION COMPLETED!")
print(f"="*80)
````

````python
print("=== APPROACH 2: HETEROGENEOUS MULTI-ALGORITHM CLUSTERING ===")
print("🔄 Creating heterogeneous teams with diverse skill sets...")

# Import additional clustering algorithms
from sklearn.cluster import AgglomerativeClustering
from sklearn.mixture import GaussianMixture

def create_heterogeneous_teams(df, features, n_teams, X_scaled_data):
    """Create heterogeneous teams using multiple clustering approaches"""
    
    # Test multiple algorithms and select the best one
    algorithms = {
        'K-Means': KMeans(n_clusters=n_teams, random_state=42),
        'Agglomerative': AgglomerativeClustering(n_clusters=n_teams),
        'Gaussian Mixture': GaussianMixture(n_components=n_teams, random_state=42)
    }
    
    best_algorithm = None
    best_score = -1
    best_labels = None
    
    # Evaluate each algorithm
    for name, algorithm in algorithms.items():
        try:
            if name == 'Gaussian Mixture':
                labels = algorithm.fit_predict(X_scaled_data)
            else:
                labels = algorithm.fit_predict(X_scaled_data)
            
            # Calculate silhouette score for evaluation
            score = silhouette_score(X_scaled_data, labels)
            print(f"   {name}: Silhouette Score = {score:.4f}")
            
            if score > best_score:
                best_score = score
                best_algorithm = name
                best_labels = labels
                
        except Exception as e:
            print(f"   {name}: Failed ({str(e)})")
    
    print(f"\n🏆 Best Algorithm: {best_algorithm} (Score: {best_score:.4f})")
    
    # Create teams with diverse skill distributions
    teams = []
    
    # Group students by cluster
    cluster_groups = {}
    for i, label in enumerate(best_labels):
        if label not in cluster_groups:
            cluster_groups[label] = []
        cluster_groups[label].append(df.iloc[i])
    
    # Create balanced teams by distributing students from different clusters
    for cluster_id in range(n_teams):
        team_students = []
        
        # Try to get students from different skill clusters
        for other_cluster_id, students in cluster_groups.items():
            if len(team_students) < 6 and len(students) > 0:
                # Select student that adds most diversity
                if len(team_students) == 0:
                    # First student - pick randomly
                    selected_student = students.pop(0)
                else:
                    # Select student that maximizes skill diversity
                    current_team_df = pd.DataFrame(team_students)
                    best_diversity = -1
                    best_student_idx = 0
                    
                    for idx, candidate in enumerate(students[:min(3, len(students))]):  # Check available candidates
                        temp_team = team_students + [candidate]
                        temp_df = pd.DataFrame(temp_team)
                        if len(temp_df) >= 2:
                            diversity = (temp_df[features].max() - temp_df[features].min()).mean()
                            if diversity > best_diversity:
                                best_diversity = diversity
                                best_student_idx = idx
                    
                    selected_student = students.pop(best_student_idx)
                
                team_students.append(selected_student)
        
        if len(team_students) >= 4:  # Only keep teams with viable size
            teams.append(pd.DataFrame(team_students))
    
    return teams, best_algorithm

# Create heterogeneous teams using the class-specific scaled data
heterogeneous_teams, best_alg = create_heterogeneous_teams(class_df, features, optimal_teams, X_scaled)

print(f"\n📊 HETEROGENEOUS TEAMS CREATED:")
print(f"   • Total Teams: {len(heterogeneous_teams)}")
print(f"   • Team Sizes: {[len(team) for team in heterogeneous_teams]}")
print(f"   • Algorithm Used: {best_alg}")

# Display team composition and diversity metrics
for i, team in enumerate(heterogeneous_teams):
    if len(team) > 0:
        print(f"\n--- Team {i+1} ({len(team)} students) ---")
        team_stats = team[features].agg(['mean', 'std']).round(2)
        print(f"Skills Distribution:")
        for skill in features:
            mean_val = team_stats.loc['mean', skill]
            std_val = team_stats.loc['std', skill]
            print(f"  {skill}: {mean_val:.2f} ± {std_val:.2f}")
        
        # Calculate diversity score
        diversity = (team[features].max() - team[features].min()).mean()
        print(f"  Diversity Score: {diversity:.2f}")

# Calculate approach metrics
diversity_scores = []
balance_scores = []

for team in heterogeneous_teams:
    if len(team) > 1:
        # Diversity: skill range within team (higher = more diverse)
        diversity = (team[features].max() - team[features].min()).mean()
        diversity_scores.append(diversity)
        
        # Balance: inverse of variance (higher = more balanced)
        team_variance = team[features].var().mean()
        balance_scores.append(1 / (1 + team_variance))

avg_diversity = np.mean(diversity_scores) if diversity_scores else 0
avg_balance = np.mean(balance_scores) if balance_scores else 0

approach2_metrics = {
    'approach': f'Heterogeneous {best_alg}',
    'num_teams': len(heterogeneous_teams),
    'avg_team_size': np.mean([len(team) for team in heterogeneous_teams]) if heterogeneous_teams else 0,
    'diversity_score': avg_diversity,
    'balance_score': avg_balance,
    'optimal_size_ratio': len([t for t in heterogeneous_teams if 4 <= len(t) <= 6]) / len(heterogeneous_teams) if heterogeneous_teams else 0
}

print(f"\n📈 APPROACH 2 METRICS:")
print(f"   • Average Diversity Score: {avg_diversity:.4f}")
print(f"   • Average Balance Score: {avg_balance:.4f}")
print(f"   • Teams in Optimal Size Range (4-6): {approach2_metrics['optimal_size_ratio']:.1%}")

print(f"\n✅ Approach 2 completed: {len(heterogeneous_teams)} heterogeneous teams created!")
````

## ⚖️ Approach 3: Hybrid Strategic Team Formation

Combine the benefits of both homogeneous and heterogeneous approaches by creating teams that balance skill similarity with strategic diversity. This approach is optimal for complex projects requiring both specialization and collaboration.

````python
print("=== APPROACH 3: HYBRID STRATEGIC TEAM FORMATION ===")
print("🔄 Creating hybrid teams with balanced specialization and diversity...")

def create_hybrid_teams(df, features, n_teams, kmeans_labels):
    """Create hybrid teams balancing homogeneity and heterogeneity"""
    
    # Step 1: Create skill-based roles
    df_copy = df.copy()
    df_copy['kmeans_cluster'] = kmeans_labels
    df_copy['primary_skill'] = df_copy[features].idxmax(axis=1)
    df_copy['skill_strength'] = df_copy[features].max(axis=1)
    df_copy['overall_score'] = df_copy[features].mean(axis=1)
    
    # Step 2: Group students by primary skills
    skill_groups = {skill: [] for skill in features}
    for _, student in df_copy.iterrows():
        primary_skill = student['primary_skill']
        skill_groups[primary_skill].append(student)
    
    print(f"\n📋 SKILL DISTRIBUTION:")
    for skill, students in skill_groups.items():
        print(f"   • {skill.replace('_', ' ').title()}: {len(students)} students")
    
    # Step 3: Create balanced teams
    teams = [[] for _ in range(n_teams)]
    
    # Distribute students strategically
    for skill_type, students in skill_groups.items():
        # Sort students by skill strength
        students.sort(key=lambda x: x['skill_strength'], reverse=True)
        
        # Distribute to teams in round-robin fashion
        for i, student in enumerate(students):
            team_idx = i % n_teams
            teams[team_idx].append(student)
    
    # Step 4: Optimize team composition
    def calculate_team_score(team_list):
        if len(team_list) < 2:
            return 0
        
        team_df = pd.DataFrame(team_list)
        
        # Balance component (lower variance = better)
        balance = 1 / (1 + team_df[features].var().mean())
        
        # Diversity component (higher range = better) 
        diversity = (team_df[features].max() - team_df[features].min()).mean() / 5.0
        
        # Size component (prefer 4-6 students)
        size_score = 1.0 if 4 <= len(team_list) <= 6 else 0.5
        
        # Skill coverage (prefer teams with different primary skills)
        unique_skills = len(team_df['primary_skill'].unique())
        coverage = unique_skills / len(features)
        
        return (balance * 0.3 + diversity * 0.3 + size_score * 0.2 + coverage * 0.2)
    
    # Optimization: swap students between teams to improve scores
    print(f"\n🔄 OPTIMIZING TEAM COMPOSITIONS...")
    
    for iteration in range(10):  # Limited iterations for efficiency
        improved = False
        
        for i in range(n_teams):
            for j in range(i + 1, n_teams):
                if len(teams[i]) > 1 and len(teams[j]) > 1:
                    # Try swapping students between teams
                    for student_i_idx in range(len(teams[i])):
                        for student_j_idx in range(len(teams[j])):
                            # Calculate current scores
                            score_i_before = calculate_team_score(teams[i])
                            score_j_before = calculate_team_score(teams[j])
                            
                            # Swap students
                            student_i = teams[i][student_i_idx]
                            student_j = teams[j][student_j_idx]
                            
                            teams[i][student_i_idx] = student_j
                            teams[j][student_j_idx] = student_i
                            
                            # Calculate new scores
                            score_i_after = calculate_team_score(teams[i])
                            score_j_after = calculate_team_score(teams[j])
                            
                            # Keep swap if it improves overall performance
                            if (score_i_after + score_j_after) > (score_i_before + score_j_before):
                                improved = True
                            else:
                                # Revert swap
                                teams[i][student_i_idx] = student_i
                                teams[j][student_j_idx] = student_j
        
        if not improved:
            break
    
    # Convert to DataFrames and filter by size
    team_dfs = []
    for team_list in teams:
        if 4 <= len(team_list) <= 6:
            team_dfs.append(pd.DataFrame(team_list))
    
    return team_dfs

# Create hybrid teams using the class-specific K-means labels
hybrid_teams = create_hybrid_teams(class_df, features, optimal_teams, class_labels)

print(f"\n📊 HYBRID TEAMS CREATED:")
print(f"   • Total Teams: {len(hybrid_teams)}")
print(f"   • Team Sizes: {[len(team) for team in hybrid_teams]}")

# Analyze team compositions
for i, team in enumerate(hybrid_teams):
    if len(team) > 0:
        print(f"\n--- Team {i+1} ({len(team)} students) ---")
        
        # Skill statistics
        team_stats = team[features].agg(['mean', 'std']).round(2)
        print(f"Skills Distribution:")
        for skill in features:
            mean_val = team_stats.loc['mean', skill]
            std_val = team_stats.loc['std', skill]
            print(f"  {skill}: {mean_val:.2f} ± {std_val:.2f}")
        
        # Team composition analysis
        primary_skills = team['primary_skill'].value_counts()
        print(f"Primary Skills Representation:")
        for skill, count in primary_skills.items():
            print(f"  {skill.replace('_', ' ').title()}: {count} students")
        
        # Performance metrics
        balance = 1 / (1 + team[features].var().mean())
        diversity = (team[features].max() - team[features].min()).mean()
        coverage = len(team['primary_skill'].unique()) / len(features)
        
        print(f"Team Metrics:")
        print(f"  Balance: {balance:.3f}, Diversity: {diversity:.3f}, Coverage: {coverage:.3f}")

# Calculate approach metrics
hybrid_balance_scores = []
hybrid_diversity_scores = []
hybrid_coverage_scores = []

for team in hybrid_teams:
    if len(team) > 1:
        # Balance score
        balance = 1 / (1 + team[features].var().mean())
        hybrid_balance_scores.append(balance)
        
        # Diversity score
        diversity = (team[features].max() - team[features].min()).mean()
        hybrid_diversity_scores.append(diversity)
        
        # Coverage score
        coverage = len(team['primary_skill'].unique()) / len(features)
        hybrid_coverage_scores.append(coverage)

approach3_metrics = {
    'approach': 'Hybrid Strategic',
    'num_teams': len(hybrid_teams),
    'avg_team_size': np.mean([len(team) for team in hybrid_teams]) if hybrid_teams else 0,
    'balance_score': np.mean(hybrid_balance_scores) if hybrid_balance_scores else 0,
    'diversity_score': np.mean(hybrid_diversity_scores) if hybrid_diversity_scores else 0,
    'coverage_score': np.mean(hybrid_coverage_scores) if hybrid_coverage_scores else 0,
    'optimal_size_ratio': len([t for t in hybrid_teams if 4 <= len(t) <= 6]) / len(hybrid_teams) if hybrid_teams else 0
}

print(f"\n📈 APPROACH 3 METRICS:")
print(f"   • Average Balance Score: {approach3_metrics['balance_score']:.4f}")
print(f"   • Average Diversity Score: {approach3_metrics['diversity_score']:.4f}")
print(f"   • Average Coverage Score: {approach3_metrics['coverage_score']:.4f}")
print(f"   • Teams in Optimal Size Range (4-6): {approach3_metrics['optimal_size_ratio']:.1%}")

print(f"\n✅ Approach 3 completed: {len(hybrid_teams)} hybrid teams created!")
````
